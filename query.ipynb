```python
import pandas as pd
from collections import Counter
import numpy as np

# Character class definition
class Character:
    def __init__(self, name, birth=None, death=None, gender=None, hair=None, height=None, race=None, realm=None, spouse=None):
        self.name = name
        self.birth = birth
        self.death = death
        self.gender = gender
        self.hair = hair
        self.height = height
        self.race = race
        self.realm = realm
        self.spouse = spouse

    def __repr__(self):
        return f"<Character {self.name}>"

    def info(self):
        return {
            "name": self.name,
            "birth": self.birth,
            "death": self.death,
            "gender": self.gender,
            "hair": self.hair,
            "height": self.height,
            "race": self.race,
            "realm": self.realm,
            "spouse": self.spouse.name if isinstance(self.spouse, Character) else self.spouse
        }

# Load characters from CSV file (comma-separated)
def load_characters(csv_file):
    df = pd.read_csv(csv_file)
    characters = {}

    for _, row in df.iterrows():
        name = row.get('name')
        if not name:
            continue
        char = Character(
            name=name,
            birth=row.get('birth'),
            death=row.get('death'),
            gender=row.get('gender'),
            hair=row.get('hair'),
            height=row.get('height'),
            race=row.get('race'),
            realm=row.get('realm'),
            spouse=row.get('spouse')
        )
        characters[name] = char

    # Link spouses as Character objects if possible
    for char in characters.values():
        spouse_name = char.spouse
        if spouse_name and spouse_name in characters:
            char.spouse = characters[spouse_name]

    return characters

# Load characters once
characters = load_characters("lotr.csv")

# Query 1: All Elves
query_output1 = pd.DataFrame(
    [char.info() for char in characters.values() if char.race and char.race.lower() == 'elf']
)
query_output1  # Display output in Jupyter

# Query 2: Count of characters by race
race_counts = Counter(char.race for char in characters.values() if char.race)
query_output2 = pd.DataFrame(race_counts.items(), columns=['Race', 'Count'])
query_output2  # Display output in Jupyter

# Query 3: Characters with valid spouses
married_characters = [
    {"name": char.name, "spouse_name": char.spouse.name}
    for char in characters.values()
    if isinstance(char.spouse, Character)
]
query_output3 = pd.DataFrame(married_characters)
query_output3  # Display output in Jupyter

# Query 4: Top 5 most common hair colors among living characters
living_chars = [char for char in characters.values() if not char.death or str(char.death).strip() == '']
hair_colors = Counter(char.hair for char in living_chars if char.hair and str(char.hair).strip())
query_output4 = pd.DataFrame(hair_colors.most_common(5), columns=['Hair Color', 'Count'])
query_output4  # Display output in Jupyter

# Query 5: Average height by gender (only numeric heights)
height_by_gender = {}
for char in characters.values():
    if char.height and char.gender:
        try:
            height_cm = float(char.height)
            gender = char.gender
            if gender not in height_by_gender:
                height_by_gender[gender] = []
            height_by_gender[gender].append(height_cm)
        except ValueError:
            pass
avg_height = {gender: np.mean(heights) for gender, heights in height_by_gender.items()}
query_output5 = pd.DataFrame(avg_height.items(), columns=['Gender', 'Average Height'])
query_output5  # Display output in Jupyter
```

```python
version:  '3'
services:
  db2:
    image: icr.io/db2_community/db2
    platform: linux/amd64
    privileged: true
    environment:
      - DB2INST1_PASSWORD=281000
      - LICENSE=accept
      - DBNAME=DB_PROJECT
    ports:
      - "50000:50000"
    volumes:
      - ./init.sql:/database/init.sql
      - ./init_db.sh:/database/init_db.sh
    command: ["/bin/bash", "-c", "/database/init_db.sh && tail -f /dev/null"]
```

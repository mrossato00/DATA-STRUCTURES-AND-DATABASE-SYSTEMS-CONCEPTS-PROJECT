```python
import ibm_db
import os

# Caricamento delle credenziali da variabili d'ambiente (fallback su valori di default)
uid = os.getenv("DB_UID", "MROSSATO00")
pwd = os.getenv("DB_PWD", "281000")

conn_str = f"""
    DATABASE=DB_PROJECT;
    HOSTNAME=localhost;
    PORT=50000;
    PROTOCOL=TCPIP;
    UID={uid};
    PWD={pwd};
"""

try:
    conn = ibm_db.connect(conn_str, "", "")
    if conn:
        server_info = ibm_db.server_info(conn)
        print(f"Connesso a {server_info.DBMS_NAME} versione {server_info.DBMS_VER}")
    else:
        print("Connessione al database fallita.")
finally:
    if 'conn' in locals() and conn:
        ibm_db.close(conn)
```

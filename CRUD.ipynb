```python
import pandas as pd

# Character class definition
class Character:
    def __init__(self, name, birth=None, death=None, gender=None, hair=None, height=None, race=None, realm=None, spouse=None):
        self.name = name
        self.birth = birth
        self.death = death
        self.gender = gender
        self.hair = hair
        self.height = height
        self.race = race
        self.realm = realm
        self.spouse = spouse  # This will be a string (name) initially

    def __repr__(self):
        return f"<Character {self.name}>"

    def info(self):
        return {
            "name": self.name,
            "birth": self.birth,
            "death": self.death,
            "gender": self.gender,
            "hair": self.hair,
            "height": self.height,
            "race": self.race,
            "realm": self.realm,
            "spouse": self.spouse
        }

# Load CSV and create characters
def load_characters(csv_file):
    df = pd.read_csv(csv_file, sep='\t')  # Assuming tab-separated; change if needed

    characters = {}

    for _, row in df.iterrows():
        name = row.get('name')
        if not name:
            continue  # Skip entries without a name

        char = Character(
            name=name,
            birth=row.get('birth'),
            death=row.get('death'),
            gender=row.get('gender'),
            hair=row.get('hair'),
            height=row.get('height'),
            race=row.get('race'),
            realm=row.get('realm'),
            spouse=row.get('spouse')
        )
        characters[name] = char

    # Optionally link spouses to Character objects instead of names
    for char in characters.values():
        spouse_name = char.spouse
        if spouse_name and spouse_name in characters:
            char.spouse = characters[spouse_name]

    return characters

# CRUD operations

def create_character(characters, **kwargs):
    name = kwargs.get('name')
    if not name:
        print("Character must have a name.")
        return None
    if name in characters:
        print(f"Character '{name}' already exists.")
        return None
    new_char = Character(**kwargs)
    characters[name] = new_char
    return new_char

def read_character(characters, name):
    return characters.get(name)

def update_character(characters, name, **kwargs):
    char = characters.get(name)
    if not char:
        print(f"No character found with name '{name}'.")
        return None
    for key, value in kwargs.items():
        if hasattr(char, key):
            setattr(char, key, value)
    return char

def delete_character(characters, name):
    if name in characters:
        del characters[name]
        # Also remove as spouse from others
        for c in characters.values():
            if isinstance(c.spouse, Character) and c.spouse.name == name:
                c.spouse = None
        return True
    else:
        print(f"No character found with name '{name}'.")
        return False

# Example usage

if __name__ == "__main__":
    characters = load_characters("lotr.csv")

    print("All characters loaded:", list(characters.keys()))

    # Read a character
    char = read_character(characters, "Boromir")
    if char:
        print("Boromir info:", char.info())

    # Create a new character
    create_character(characters, name="Frodo", race="Hobbit", gender="Male")

    # Update a character
    update_character(characters, "Frodo", hair="Brown", height="Short")

    # Delete a character
    delete_character(characters, "Angrim")

    # Check updates
    print(characters.get("Frodo").info())
```
